```python
# next I'll group the apps based on install ranges (1K-10K, 100K-500K)
# and calculate the corresponding app counts.
# set the bins
bins = [
    0,
    1000,
    10000,
    100000,
    500000,
    1000000,
    5000000,
    10000000,
    50000000,
    100000000,
    float('inf')]
# set the labels
labels = [
    '0-1,000',
    '1,001-10,000',
    '10,001-100,000',
    '100,001-500,000',
    '500,001-1,000,000',
    '1,000,001-5,000,000',
    '5,000,001-10,000,000',
    '10,000,001-50,000,000',
    '50,000,001-100,000,000',
    '100,000,001+']
    

```

```python
# Create the ranges
google_data['Install Range'] =pd.cut(
    google_data['Installs'],
    bins=bins,
    labels=labels,
    include_lowest=True)
```

```python
# Count the apps
install_counts = google_data['Install Range'].value_counts().sort_index()
plt.figure(figsize=(12, 6))

ax = install_counts.plot(kind='bar')

plt.title('Distribution of Apps by Install Range')
plt.xlabel('Install Range')
plt.ylabel('Number of Apps')
plt.xticks(rotation=45)
for i, value in enumerate(install_counts):
    ax.text(i, value, str(value), ha='center', va='bottom')
    plt.tight_layout()
    plt.show()
    
```

```python
print(install_counts)
```

```python
# Here I'll count apps by install range and category
category_counts = google_data.groupby(
    ['Install Range', 'Category'],
    observed=True
).size().reset_index(name='App Count')

print(category_counts)
    
    
    
```

```python
# Overall most popular install ranges
most_popular = category_counts.loc[
    category_counts.groupby('Install Range')['App Count'].idxmax()]
print(most_popular)
```

```python
# Find the overall most popular install ranges
range_totals = google_data['Install Range'].value_counts().sort_values(
    ascending=False)
print(range_totals)
```

```python
# The combined results into one table
result = most_popular.sort_values(
    'App Count',
    ascending=False)
print(result)
```
