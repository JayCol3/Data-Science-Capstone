# Identifying missing values, Ratings i'll fill with the mean, and drop version, content rating, current version and type.

```python
google_data['Rating'] = google_data['Rating'].fillna(google_data['Rating'].median())
```

```python
google_data['Type'] = google_data['Type'].fillna(google_data['Type'].mode()[0])
```

```python
google_data['Content Rating'] = google_data['Content Rating'].fillna(google_data['Content Rating'].mode()[0])
```

```python
# Current version , not numeric, mode isn't meaningful, safest thing to do there's no way to compute versions
google_data = google_data.dropna(subset=['Current Ver'])
```

```python
# We'll also drop Android  version not enough to make a difference
google_data = google_data.dropna(subset=['Android Ver'])
```
